### PEAK HOURS FOR ASB INCIDENTS

```python
df['Hour'] = pd.to_datetime(df['Hour'], format='%H:%M').dt.hour
```

### RESPONSE TIME ANALYSIS

```python
df['Response_Time'] = pd.to_numeric(df['Response_Time'], errors='coerce')
```

```python
response_by_borough = df.groupby('Safer_Neighborhood_Team_Borough_Name')['Response_Time'].mean().sort_values()

# Reset index for plotting
response_df = response_by_borough.reset_index()
response_df.columns = ['Borough', 'Average_Response_Time']
```

```python
top5_fastest = response_df.head(5)
top5_slowest = response_df.tail(5)
combined_df = pd.concat([top5_fastest, top5_slowest])
```

```python
plt.figure(figsize=(12, 6))
ax = sns.barplot(data=combined_df, x='Borough', y='Average_Response_Time')
ax.bar_label(ax.containers[0], fmt='%.1f', rotation=90)

# Labels and title
plt.title('Top 5 Fastest and Slowest Boroughs by Police Response Time')
plt.xlabel('Borough')
plt.ylabel('Average Response Time (Secs)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Remove outliers using IQR
Q1 = df['Response_Time'].quantile(0.25)
Q3 = df['Response_Time'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df['Response_Time'] >= lower_bound) & (df['Response_Time'] <= upper_bound)]
```

```python
# Filter relevant ASB types
asb_types = ['ASB Personal', 'ASB Nuisance', 'ASB Environmental Substance Misuse']
df = df[df['Opening_Type_1'].isin(asb_types)]

# ✅ Boxplot: Response Time by ASB Type
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="Opening_Type_1", y="Response_Time", palette="Set2",hue="Opening_Type_1", legend= False)
plt.title("Response Time by ASB Type")
plt.ylabel("Response Time (minutes)")
plt.xlabel("ASB Type")
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
# Find duplicates using Borough, ASB Type, and Response Time
df['Duplicate'] = df.duplicated(subset=['Safer_Neighborhood_Team_Borough_Name', 'Opening_Type_1', 'Response_Time'], keep=False)

# Count duplicates by borough
duplicate_counts = df[df['Duplicate']].groupby('Safer_Neighborhood_Team_Borough_Name').size().reset_index(name='Duplicate_Count')

# Sort the data
duplicate_counts = duplicate_counts.sort_values(by='Duplicate_Count', ascending=False)

# Bar chart
plt.figure(figsize=(12, 6))
sns.barplot(data=duplicate_counts, x='Safer_Neighborhood_Team_Borough_Name', y='Duplicate_Count',  hue='Safer_Neighborhood_Team_Borough_Name', palette='crest', legend=False)
plt.title('Duplicate ASB Reports by Borough (Estimated)')
plt.xticks(rotation=45, ha='right')
plt.xlabel('Borough')
plt.ylabel('Number of Duplicate Reports')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
